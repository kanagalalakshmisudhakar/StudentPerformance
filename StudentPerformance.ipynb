## Prepare Data for Machine Learning

### Subtask:
Prepare the dataset for machine learning by separating features (X) and target (y), encoding categorical variables, and splitting the data into training and testing sets.


**Reasoning**:
The subtask requires separating features (X) and target (y), encoding categorical variables, and splitting the data. I will start by defining the target and features, then identify and one-hot encode the categorical columns in the features.



````python
from sklearn.model_selection import train_test_split
import pandas as pd

# 1. Define target (y) and features (X)
y = df['G3']
X = df.drop('G3', axis=1)

# 2. Identify categorical columns in X
categorical_cols = X.select_dtypes(include=['object']).columns

# 3. Apply one-hot encoding to categorical columns
X_encoded = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

# Display the shape of the encoded features and the target
print(f"Shape of features (X_encoded): {X_encoded.shape}")
print(f"Shape of target (y): {y.shape}")

# Display the first few rows of the encoded features
print("\nFirst 5 rows of X_encoded:")
display(X_encoded.head())
````

**Reasoning**:
The previous step defined the target and features and encoded the categorical variables. Now, I will split the preprocessed data (`X_encoded` and `y`) into training and testing sets using `train_test_split` with a `test_size` of 0.2 and a `random_state` for reproducibility, as per the subtask instructions.



````python
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

print(f"\nShape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")
````

**Reasoning**:
The subtask requires training a Random Forest Regressor model and evaluating its performance. I will import the necessary libraries, instantiate the model, train it on the prepared training data, make predictions on the test set, and then calculate and print the MAE, MSE, and R2 scores.



````python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Instantiate a RandomForestRegressor model
model = RandomForestRegressor(random_state=42)

# 2. Train the model
model.fit(X_train, y_train)

# 3. Make predictions on the test set
y_pred = model.predict(X_test)

# 4. Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# 5. Print the calculated metrics
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"R-squared (R2): {r2:.2f}")
````

## Build Machine Learning Model

### Subtask:
Train a Random Forest Regressor model to predict the final grade (G3) and evaluate its performance using Mean Absolute Error (MAE), Mean Squared Error (MSE), and R-squared (R2).


**Reasoning**:
I will create a Gradio interface to deploy the trained Random Forest Regressor model. This involves defining a prediction function that preprocesses input data similarly to how the training data was processed, creating appropriate Gradio input components for each feature, and then launching the Gradio interface.



````python
import gradio as gr
import numpy as np

# Get the list of original categorical and numerical columns from df for Gradio inputs
categorical_cols_original = df.select_dtypes(include=['object']).columns.tolist()
numerical_cols_original = df.select_dtypes(include=['int64', 'float64']).drop(columns=['G3', 'G1', 'G2'], errors='ignore').columns.tolist()

def predict_g3(*args):
    # Reconstruct the input features into a dictionary
    input_data = dict(zip(X.columns, args))

    # Create a DataFrame from the input data
    input_df = pd.DataFrame([input_data])

    # Apply one-hot encoding to categorical columns, consistent with training data
    # Use the categorical columns identified from the original X before encoding
    categorical_cols_for_encoding = X.select_dtypes(include=['object']).columns
    input_df_encoded = pd.get_dummies(input_df, columns=categorical_cols_for_encoding, drop_first=True)

    # Ensure all columns from X_encoded are present, fill missing with 0
    # This is crucial because get_dummies might not produce all columns if a category is missing in the input
    for col in X_encoded.columns:
        if col not in input_df_encoded.columns:
            input_df_encoded[col] = 0

    # Reorder columns to match the training data's column order
    input_df_encoded = input_df_encoded[X_encoded.columns]

    # Make prediction
    prediction = model.predict(input_df_encoded)[0]

    return f"{prediction:.2f}"

# Create Gradio input components
gradio_inputs = []
for col in X.columns:
    if col in df.select_dtypes(include=['object']).columns:
        # Categorical feature: use Dropdown
        choices = df[col].unique().tolist()
        gradio_inputs.append(gr.Dropdown(choices, label=col))
    elif col in df.select_dtypes(include=['int64', 'float64']).columns:
        # Numerical feature: use Slider (assuming integer steps for simplicity)
        min_val = int(df[col].min())
        max_val = int(df[col].max())
        gradio_inputs.append(gr.Slider(minimum=min_val, maximum=max_val, step=1, label=col))

# Create the Gradio Interface
iface = gr.Interface(
    fn=predict_g3,
    inputs=gradio_inputs,
    outputs=gr.Label(label="Predicted Final Grade (G3)"),
    title="Student Performance Predictor",
    description="Predict the final grade (G3) based on various student and family attributes."
)

# Launch the interface
iface.launch(share=True)
````
